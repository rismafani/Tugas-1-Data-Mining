**Membersihkan Format dan Menyeragamkan Data**

```python

# Membuat salinan dataset
df_clean = df.copy()

# 1. Membersihkan Jenis Kelamin
df_clean['Jenis_Kelamin'] = (
    df_clean['Jenis_Kelamin'].astype('string').str.strip().str.lower()
)

df_clean['Jenis_Kelamin'] = df_clean['Jenis_Kelamin'].replace({
    'l': 'Laki-laki',
    'pria': 'Laki-laki',
    'laki-laki': 'Laki-laki',
    'perempuan': 'Perempuan'
})

# 2. Membersihkan Kota
df_clean['Kota'] = (
    df_clean['Kota'].astype('string').str.strip().str.lower()
)

df_clean['Kota'] = df_clean['Kota'].replace({
    'surabya': 'Surabaya',
    'surabaya': 'Surabaya',
    'malang': 'Malang',
    'sidoharjo': 'Sidoarjo',
    'sidoarjo': 'Sidoarjo',
    'gresik': 'Gresik',
    'mojokerto': 'Mojokerto'
})

# 3. Membersihkan Metode Pembayaran
df_clean['Metode_Pembayaran'] = (
    df_clean['Metode_Pembayaran'].astype('string').str.strip().str.lower()
)

df_clean['Metode_Pembayaran'] = df_clean['Metode_Pembayaran'].replace({
    'cod': 'COD',
    'e-wallet': 'E-Wallet',
    'e wallet': 'E-Wallet',
    'ewallet': 'E-Wallet',
    'transfer': 'Transfer Bank',
    'transfer bank': 'Transfer Bank',
    'kartu kredit': 'Kartu Kredit'
})

# 4. Mengubah Usia menjadi numerik
df_clean['Usia'] = pd.to_numeric(
    df_clean['Usia'].astype('string').str.extract(r'(-?\d+\.?\d*)')[0],
    errors='coerce'
)

# 5. Membersihkan format Rupiah
def bersihkan_rupiah(nilai):
    if pd.isna(nilai):
        return np.nan

    nilai = str(nilai).strip()

    if 'Rp' in nilai or (
        nilai.count('.') > 1
    ):
        nilai = ''.join(filter(str.isdigit, nilai))
        return float(nilai) if nilai else np.nan

    return pd.to_numeric(nilai, errors='coerce')

df_clean['Pendapatan_Bulanan'] = df_clean[
    'Pendapatan_Bulanan'
].apply(bersihkan_rupiah)

df_clean['Total_Belanja'] = df_clean[
    'Total_Belanja'
].apply(bersihkan_rupiah)

# 6. Menampilkan hasil
print(df_clean.head(10))
print("\nTipe Data Setelah Pembersihan:")
print(df_clean.dtypes)
```

**Menghitung Statistik Deskriptif Kembali**

```python

kolom_numerik = [
    'Usia',
    'Pendapatan_Bulanan',
    'Frekuensi_Belanja',
    'Total_Belanja',
    'Rating_Kepuasan'
]

print(df_clean[kolom_numerik].describe())
```

**Memeriksa Nilai yang Tidak Wajar**

```python

# Memeriksa nilai yang tidak valid
print("Usia di luar rentang 1-100:")
print(df_clean[
    (df_clean['Usia'] < 1) | (df_clean['Usia'] > 100)
][['ID_Pelanggan', 'Usia']])

print("\nRating di luar skala 1-5:")
print(df_clean[
    (df_clean['Rating_Kepuasan'] < 1) |
    (df_clean['Rating_Kepuasan'] > 5)
][['ID_Pelanggan', 'Rating_Kepuasan']])

print("\nFrekuensi belanja di atas 100:")
print(df_clean[
    df_clean['Frekuensi_Belanja'] > 100
][['ID_Pelanggan', 'Frekuensi_Belanja']])
```

```python

# 1. Menangani usia tidak valid
df_clean.loc[
    (df_clean['Usia'] < 1) | (df_clean['Usia'] > 100),
    'Usia'
] = np.nan

# 2. Menangani rating tidak valid
df_clean.loc[
    (df_clean['Rating_Kepuasan'] < 1) |
    (df_clean['Rating_Kepuasan'] > 5),
    'Rating_Kepuasan'
] = np.nan

# 3. Menangani pendapatan negatif
df_clean.loc[
    df_clean['Pendapatan_Bulanan'] < 0,
    'Pendapatan_Bulanan'
] = np.nan

# 4. Menangani frekuensi belanja ekstrem
df_clean.loc[
    df_clean['Frekuensi_Belanja'] > 100,
    'Frekuensi_Belanja'
] = np.nan

print("Statistik setelah menangani nilai tidak valid:")
print(df_clean[kolom_numerik].describe())

print("\nJumlah missing value terbaru:")
print(df_clean.isnull().sum())
```

**Menangani Outlier**

```python

# Memeriksa outlier menggunakan metode IQR
for kolom in [
    'Usia',
    'Pendapatan_Bulanan',
    'Frekuensi_Belanja',
    'Total_Belanja',
    'Rating_Kepuasan'
]:
    Q1 = df_clean[kolom].quantile(0.25)
    Q3 = df_clean[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    jumlah_outlier = (
        (df_clean[kolom] < batas_bawah) |
        (df_clean[kolom] > batas_atas)
    ).sum()

    print(f"\n{kolom}")
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)
    print("Jumlah outlier:", jumlah_outlier)
```

**Melihat Data Outlier**

```python

# Memeriksa data yang terdeteksi sebagai outlier
kolom_outlier = [
    'Pendapatan_Bulanan',
    'Frekuensi_Belanja',
    'Total_Belanja'
]

for kolom in kolom_outlier:
    Q1 = df_clean[kolom].quantile(0.25)
    Q3 = df_clean[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    kondisi = (
        (df_clean[kolom] < batas_bawah) |
        (df_clean[kolom] > batas_atas)
    )

    print(f"\nData outlier pada {kolom}:")
    print(df_clean.loc[
        kondisi, ['ID_Pelanggan', kolom]
    ])
```

**Menangani Outlier dengan Capping**

```python

# Menangani outlier menggunakan metode IQR Capping

kolom_outlier = [
    'Pendapatan_Bulanan',
    'Frekuensi_Belanja',
    'Total_Belanja'
]

for kolom in kolom_outlier:
    Q1 = df_clean[kolom].quantile(0.25)
    Q3 = df_clean[kolom].quantile(0.75)
    IQR = Q3 - Q1

    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR

    df_clean[kolom] = df_clean[kolom].clip(
        lower=batas_bawah,
        upper=batas_atas
    )

    print(f"\n{kolom}")
    print("Batas bawah:", batas_bawah)
    print("Batas atas:", batas_atas)

print("\nStatistik setelah capping:")
print(df_clean[kolom_numerik].describe())
```

**Mengisi missing value numerik menggunakan median**

```python

# Mengisi missing value numerik dengan median

kolom_numerik = [
    'Usia',
    'Pendapatan_Bulanan',
    'Frekuensi_Belanja',
    'Total_Belanja',
    'Rating_Kepuasan'
]

for kolom in kolom_numerik:
    median = df_clean[kolom].median()
    df_clean[kolom] = df_clean[kolom].fillna(median)

    print(f"{kolom}: Missing value diisi median = {median}")
```

**Mengisi missing value kategorik menggunakan modus**

```python

# Mengisi missing value kategorik dengan modus

kolom_kategorik = [
    'Jenis_Kelamin',
    'Kota',
    'Metode_Pembayaran',
    'Kategori_Favorit'
]

for kolom in kolom_kategorik:
    modus = df_clean[kolom].mode()[0]
    df_clean[kolom] = df_clean[kolom].fillna(modus)

    print(f"{kolom}: Missing value diisi modus = {modus}")
```
